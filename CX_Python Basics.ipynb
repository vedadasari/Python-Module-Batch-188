string functions

lower(): converts the string into lowercase
upper(): converts the string into uppercase
title(): converts first letter of the string to uppercase and others will be lowercase

```python
"PUNE".lower()=="pune"
```

```python
"Pune".lower()=="pune"
```

```python
"Pune".lower()
```

```python
location3=input("Enter your location")
if (age>=18) and location3.lower() in ("pune","kolhapur","nashik","aurangaba"):
    print("Eligible")
else:
    print("Not Eligible")
```

```python
location3 = input("Enter the location details")
if (age>=18) and location3.upper() in ("PUNE","NASHIK","AURANGABAD","MUMBAI","KOLHAPUR"):
    print("Eligible")
else:
    print("Not Eligible")
```

```python
#title makes the first letter to upper case of the word
location3 = input("Enter the location details")
if (age>=18) and location3.title() in ("Pune","Nashik","Aurangabad","Mumbai","Kolhapur"):
    print("Eligible")
else:
    print("Not Eligible")
```

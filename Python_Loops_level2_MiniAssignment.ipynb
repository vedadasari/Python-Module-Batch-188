1. List – Positive, Negative and Zero Classification
Given a list of integers, write a program to:
• Count positive numbers, negative numbers and zeros separately.
• Display the count of each category.
• Calculate the sum of positive numbers only.
numbers = [10, -5, 0, 25, -12, 8, 0, -3, 15]



```python
numbers = [10, -5, 0, 25, -12, 8, 0, -3, 15]
positive_numbers=[]
negative_numbers=[]
zero=[]
for n in numbers:
    if n==0:
        zero.append(n)
    elif n>0:
        positive_numbers.append(n)
    else:
        negative_numbers.append(n)
#print(numbers,positive_numbers,negative_numbers,zero,sep='\n')
print("Count of positive numbers:",len(positive_numbers))
print("Count of negative numbers:",len(negative_numbers))
print("Count of zeroes:",len(zero))
print("Sum of positive numbers:",sum(positive_numbers))
```

2. List – Find Numbers That Occur at Prime Index Positions
Given a list of integers, display the elements that are located at prime index positions (2, 3, 5, 7, ...). Also calculate the sum of those selected elements. Treat the first element as index 0.
numbers = [14, 21, 8, 17, 30, 11, 25, 9, 40]


```python
numbers1 = [14, 21, 8, 17, 30, 11, 25, 9, 40]
prime_index=[2,3,5,7]
selected_numbers =[]
for i in prime_index:
    if i<len(numbers1):
        selected_numbers.append(numbers1[i])

print("Elements at prime index positions:",selected_numbers)
print("Sum of selected numbers:",sum(selected_numbers))
```

3. List – Identify Local Peaks
A local peak is an element that is greater than both its immediate neighbours. Traverse the list and display all local peak values. Do not consider the first and last elements because they do not have two neighbours.
numbers = [4, 9, 3, 12, 7, 15, 10, 6, 11, 2]


```python
numbers2 = [4, 9, 3, 12, 7, 15, 10, 6, 11, 2]
peaks=[]
for i in range(1,len(numbers2)-1):
    if numbers2[i]>numbers2[i-1] and numbers2[i]>numbers2[i+1]:
        peaks.append(numbers2[i])
print("Local Peaks:",peaks)
```

Ignore numbers that are not divisible by either.
numbers = [6, 8, 9, 12, 15, 20, 24, 25, 30]

5. String – Build a Character Frequency Report
Given a string, count the frequency of each distinct character using loops. Display each character only once along with its frequency. Ignore spaces and treat uppercase and lowercase letters as the same character.
text = "Data Science is interesting"


```python
text = "Data Science is interesting".lower()
frequency =""
for t in text:
    if t!=" " and t not in frequency:
        count = 0
        for c in text:
            if c == t:
                count += 1

        print(t, "=", count)
        frequency += t
```

6. String – Palindrome Checker
Write a program to check whether a given string is a palindrome without using string slicing. Ignore spaces and uppercase/lowercase differences.


```python
text1 = input("Enter a string: ")
# Convert to lowercase and remove spaces
text1 = text1.lower().replace(" ", "")

reverse_text = ""

for ch in text1:
    reverse_text = ch + reverse_text

if text1 == reverse_text:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

7. String – Longest Word Finder
Given a sentence, identify the longest word without using the max() function. If two or more words have the same maximum length, display all of them.
sentence = "Python programming develops logical thinking"


```python
sentence = "Python programming develops logical thinking"
words = sentence.split()
max_length = 0
longest_words = []

for w in words:
    if len(w) > max_length:
        max_length = len(w)
        longest_words = [w]
    elif len(w) == max_length:
        longest_words.append(w)

print("Longest word(s):", longest_words)
print("Length:", max_length)
```

8. Tuple – Product Sales Summary
A tuple contains the number of units sold for different products over one day. Display the highest and lowest sales values using loops (do not use max() or min()). Also count how many products sold more than 10 units and how many sold fewer than 5 units.
sales = (8, 15, 3, 22, 11, 4, 18, 6, 2)


```python
sales = (8, 15, 3, 22, 11, 4, 18, 6, 2)
highest = sales[0]
lowest = sales[0]
more_than_10 = 0
less_than_5 = 0

for s in sales:
    if s > highest:
        highest = s

    if s < lowest:
        lowest = s

    if s > 10:
        more_than_10 += 1

    if s < 5:
        less_than_5 += 1

print("Highest sales:", highest)
print("Lowest sales:", lowest)
print("Products sold more than 10 units:", more_than_10)
print("Products sold fewer than 5 units:", less_than_5)
```

9. Tuple – Identify Duplicate Elements
Given a tuple, identify all elements that appear more than once. Do not convert the tuple into a set or use count(). Display each duplicate element only once.
numbers = (10, 20, 30, 20, 40, 10, 50, 30, 60)


```python
numbers4 = (10, 20, 30, 20, 40, 10, 50, 30, 60)
duplicates = []

for num in numbers4:
    freq = 0

    for item in numbers4:
        if num == item:
            freq += 1

    if freq > 1 and num not in duplicates:
        duplicates.append(num)

print(duplicates)
```

10. List – Simple Inventory Management
A shop maintains product quantities in a list. Write a program to:
• Display products that are out of stock (quantity = 0).
• Display products with low stock (quantity between 1 and 5).
• Display products with sufficient stock (quantity greater than 5).
• Count the number of products in each category.
• Calculate the total available inventory (sum of all quantities).
quantities = [5, 0, 12, 3, 0, 8, 2, 15]

```python
quantities = [5, 0, 12, 3, 0, 8, 2, 15]
out_count = low_count = sufficient_count = 0
total = 0

for q in quantities:
    total += q

    if q == 0:
        print(q, "- Out of Stock")
        out_count += 1

    elif q <= 5:
        print(q, "- Low Stock")
        low_count += 1

    else:
        print(q, "- Sufficient Stock")
        sufficient_count += 1

print("\nOut of Stock Count:", out_count)
print("Low Stock Count:", low_count)
print("Sufficient Stock Count:", sufficient_count)
print("Total Inventory:", total)
```
